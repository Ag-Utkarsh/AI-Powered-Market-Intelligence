# Task
Create a dataframe by iterating through 20 unique values of the 'Category' column from the dataframe `googledf`, using each category as a query parameter in a RapidAPI call with `num=50`. Implement rate limiting and retry logic for the API calls. The resulting dataframe should have the same columns as the original `googledf` dataframe.

## Define the api call function

### Subtask:
Create a function to make API calls with rate limiting and retry logic.


**Reasoning**:
Define a function to handle API calls with rate limiting and retry logic as per the instructions.



```python
import time

def call_app_store_api(category, api_key, num_results=50, max_retries=3, initial_delay=1):
    """
    Makes an API call to the App Store scrapper API with rate limiting and retry logic.

    Args:
        category (str): The app category to search for.
        api_key (str): Your RapidAPI key.
        num_results (int): The number of results to request (default is 50).
        max_retries (int): The maximum number of retries for API calls (default is 3).
        initial_delay (int): The initial delay in seconds before retrying (default is 1).

    Returns:
        list: A list of dictionaries containing the API response data, or an empty list if the call fails after retries.
    """
    base_url = "https://appstore-scrapper-api.p.rapidapi.com/v1/app-store-api/search"
    url = f"{base_url}?num={num_results}&lang=en&query={category}&country=us"
    headers = {
        "X-RapidAPI-Key": api_key,
        "X-RapidAPI-Host": "appstore-scrapper-api.p.rapidapi.com"
    }

    for retry_count in range(max_retries):
        try:
            response = requests.get(url, headers=headers)
            response.raise_for_status()  # Raise an HTTPError for bad responses (4xx or 5xx)

            if response.status_code == 200:
                return response.json()

        except requests.exceptions.HTTPError as e:
            if response.status_code == 429:
                print(f"Rate limit exceeded for category '{category}'. Retrying in {initial_delay * (2**retry_count)} seconds...")
                time.sleep(initial_delay * (2**retry_count))
            else:
                print(f"HTTP error for category '{category}': {e}")
                break
        except requests.exceptions.RequestException as e:
            print(f"Request error for category '{category}': {e}")
            break

    print(f"Failed to retrieve data for category '{category}' after {max_retries} retries.")
    return []
```

## Iterate through categories and fetch data

### Subtask:
Loop through the unique categories, call the API for each category using the function created in step 1, and collect the results.


**Reasoning**:
I need to iterate through the first 20 unique categories from `googledf`, call the API function for each, and collect the results in a list.



```python
unique_categories = googledf['Category'].unique()
selected_categories = unique_categories[:20]
all_app_data = []

# Replace with your actual RapidAPI key
rapidapi_key = ""

for category in selected_categories:
    print(f"Fetching data for category: {category}")
    category_data = call_app_store_api(category, rapidapi_key, num_results=50)
    all_app_data.extend(category_data)

print(f"Collected data for {len(all_app_data)} apps.")
```
